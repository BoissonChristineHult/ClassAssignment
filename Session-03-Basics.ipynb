```python
def count_hashtags_starting_with(hashtags, letter):
    """Count how many hashtags in the list start with the given letter, case insensitive."""
    count = 0
    letter_lower = letter.lower()
    for hashtag in hashtags:
        # remove the # symbol and check the first real letter
        first_char = hashtag[1].lower()
        if first_char == letter_lower:
            count = count + 1
    return count

# Example
my_hashtags = ['#travel', '#food', '#technology', '#sunset', '#tea']
result = count_hashtags_starting_with(my_hashtags, 'T')
print(result)
```

```python
#Prompt used: [Write me a function that takes a list of hashtags and a letter, and counts how many hashtags start with that letter, ignoring whether its upper or lowercase]
AI used: Claude Sonnet 4.5 on claude.ai
```

```python
The function loops through every hashtag in the list and checks the character at position 1 (skipping the # symbol at position 0). It converts both the hashtag's first letter and the search letter to lowercase before comparing, so the search works regardless of capitalisation. Every time it finds a match it adds 1 to a counter, and returns the final count at the end.
```

```python
my_hashtags = ['#sunshine', '#summer', '#food', '#fitness', '#sunset', '#travel']
result = count_hashtags_starting_with(my_hashtags, 'S')
print(result)
```

```python
# List of likes per post
likes_per_post = [120, 340, 89, 512, 275, 410]

# Track the highest like count and its position
highest_likes = 0
best_post_position = 0

# Loop through the list and check each one
for i in range(len(likes_per_post)):
    if likes_per_post[i] > highest_likes:
        highest_likes = likes_per_post[i]
        best_post_position = i + 1  # convert from 0-based to 1-based

print('Post', best_post_position, 'was the top performer with', highest_likes, 'likes')
```
